```python
# Basic Modules
import pandas as pd
import numpy as np
import scipy as stats
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load cleaned data files into pandas DataFrames
# 'ISO-8859-1' encoding is used for building_descriptions_df to handle special characters
# that would cause issues with the default 'utf-8' encoding

yearly_values_df = pd.read_csv("Cleaned Data\cleaned_yearly_values.csv")
sales_data_df = pd.read_csv("Cleaned Data\cleaned_sales_date.csv")
building_descriptions_df = pd.read_csv("Cleaned Data\cleaned_building_descriptions.csv", encoding='ISO-8859-1')
```

```python
num_cols = yearly_values_df.select_dtypes(include=['number']).columns

# Step 4: Plot each numerical column (histogram or boxplot)
for col in num_cols:
    plt.figure(figsize=(8, 4))
    sns.distplot(yearly_values_df[col].dropna(), kde=True, bins=30)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
